Solution to [Day 3 Lobby, Batteries and Joltages](https://adventofcode.com/2025/day/2).

Part 1. Finding the maximum 2-digit number from a string of digits. This uses the function *get_highest_num*.

Part 2. Finding the maximum 12-digit number from a string of digits. This uses a loop and a dynamic function *get_highest_num_pt2*, in which the number of digits can be changed. 

In both parts, the sum of the maximum numbers are added up to give the final answer. 

-Annette

```python
def get_highest_num(numstr:str) -> int:
    '''Returns the highest two-digit number of a string of digits'''
    firstdigit=max(numstr[:-1])
    seconddigit = max(numstr[numstr.index(firstdigit)+1:])
    return int(firstdigit + seconddigit)
    
```

```python
#Part 1 Answer

with open('inputs/day3_input.txt','r') as f:
    lines = f.read().splitlines()

print(f'Part 1 total of highest numbers is {sum([get_highest_num(numstr) for numstr in lines])}')

```

```python
def get_highest_num_pt2(numstr:str, n=12) -> int:
    '''Returns the highest n-digit number of a string of digits.
    Default value of n=12 for this task.'''
    max_str = ''
    for i in range(n-1,0,-1): #decreasing range, gives first n-1 digits
        next_digit = max(numstr[:-i])
        max_str += next_digit
        numstr = numstr[numstr.index(next_digit)+1:]
    max_str += max(numstr) #gives last digit
    return int(max_str)
```

```python
#Part 2 Answer

print(f'Part 2 total of highest numbers is {sum([get_highest_num_pt2(numstr) for numstr in lines])}')
```
