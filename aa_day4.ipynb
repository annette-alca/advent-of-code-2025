Solution to [Day 4 Printer Rolls and Forklifts at the Printing Department](https://adventofcode.com/2025/day/4).

The first 2-dimensional matrix challenge of AoC 2025. 

Part 1. Find total number of printer rolls with 3 or less adjacent rolls in the matrix. The functions *count_printer_rolls* and *minesweep* are refactored from similar challenges from previous years.

Part 2. The identified rolls are removed by a forklift. And the process starts again to find new rolls meeting the criteria in Part 1, then remove them. These steps are repeated until no more rolls meet the criteria. A perfect case for recursion. The recursive function *clear_and_count_rolls* uses the two Part 1 functions and *clear_rolls* until the end condition is reached. I present its usage below, as well as a second solution that does not use recursion.

```python
#Part 1 and Part 2 Functions

def minesweep (mtx: list[str], row: int, col: int, maxrow: int, maxcol: int) -> bool:
    '''count rolls around a coordinate.
    if it is 4 or less (including the coordinate), return True
    otherwise, return False
    '''
    roll_cluster_count = 0
    for rowval in [row-1,row, row+1]:
        if rowval>=0 and rowval<maxrow:
            for colval in [col-1,col, col+1]:
                if colval>=0 and colval <maxcol:
                    if mtx[rowval][colval] == '@':
                        roll_cluster_count +=1
    # print(row,col, roll_cluster_count)
    return roll_cluster_count<=4 #


def count_printer_rolls (mtx:list[str]) -> int:
    '''Returns number of valid rolls'''
    maxrow = len(mtx)
    maxcol = len(mtx[0])
    
    validrolls = 0 #total that meet minesweep conditions
    roll_coords = [] #coordinates of validrolls 
    
    for row in range(maxrow):
        for col in range(maxcol):

            if not mtx[row][col] == '@':
                continue
            else:
                if minesweep(mtx, row, col, maxrow, maxcol):
                    validrolls += 1
                    roll_coords.append((row,col))
    return validrolls, roll_coords
```

```python
#Part 1 Sample

filename = 'inputs/day4_sample.txt'

with open(filename,'r') as f:
    sample = f.read().splitlines()

count_printer_rolls(sample)[0]
```

```python
#Part 1 Answer

filename = 'inputs/day4_input.txt'

with open(filename,'r') as f:
    matrix = f.read().splitlines()

validrolls, roll_coords = count_printer_rolls(matrix)
print(f"Part 1. Total number of rolls that can be removed = {validrolls}.")
```

```python
# Part 2 Functions

def clear_rolls(mtx: list[str], roll_coords: list[(int,int)]) -> list[str]:
    '''Modifies matrix after removing list of coordinates.
    Modifies in place. Returns None'''
    for row,col in roll_coords:
        rowstr = mtx[row]
        mtx[row] = rowstr[:col] + 'x' + rowstr[col+1:]

def clear_and_count_rolls(mtx: list[str], total_rolls_removable: int = 0) -> int:
    '''A recursive function that uses Part 1 functions and clear_rolls to find all forklift-removable rolls.
    Returns the total number'''
    validrolls, roll_coords = count_printer_rolls(mtx)
    total_rolls_removable += validrolls
    clear_rolls(mtx, roll_coords)
    if validrolls == 0:
        return total_rolls_removable
    else:
        return clear_and_count_rolls(mtx, total_rolls_removable)
```

```python
# Part 2 Sample

mtx = sample.copy()
print(clear_and_count_rolls(mtx))

```

```python
# Part 2 Solution

mtx = matrix.copy()
print(f"Part 2 using recursive function. Total rolls that can be removed = {clear_and_count_rolls(mtx)}.")
```

```python
# Part 2 Solution Without Recursion

mtx = matrix.copy()
total_rolls_removable = 0
validrolls = 1 #placeholder value
while validrolls>0:
    validrolls, roll_coords = count_printer_rolls(mtx)
    total_rolls_removable += validrolls
    clear_rolls(mtx, roll_coords)
print(f"Part 2 No recursion. Total rolls that can be removed = {total_rolls_removable}.")
```
