Solution to [Day 7 Laboratories. Tachyon Manifold Beams](https://adventofcode.com/2025/day/7).

Part 1. Count the number of times the beam is split down the path. Part 1 uses *new_beams_per_line* to keep track of beam locations and the number of times a beam is split for each new line (or row) of data. The function *count_all_splits* loops through the data to sum up all the splits.

Part 2. Single particle, not a beam! Count the many possible routes for this particle. I refactored the Part 1 functions; *new_particle_routes_per_line* creates a dictionary for each line (or row) of data, all possible particle locations : number of routes so far through that location. The function *count_all_routes* loops through the data, creating a new dictionary at every row. At the end of the loop, all dictionary values are added up, representing the total of all possible routes throughout the manifold.

```python
#Part 1 Functions

def new_beams_per_line(newline:str, beam_locs:set[int]) -> (int, set[int]):
    '''Helper function for Part 1 task.
    newline : str of new line with splittes ^ and spaces .
    beam_locs: set of integers, using sets to give only unique indices
    Returns: hit_splitters : int, number of splitters the incoming beams hit
        new_locs: set of integers, unique indices of outgoing beams
        ''' 
    hit_splitters = 0
    new_locs = set[int]()
    for indy in beam_locs: #indices of incoming beams
        if newline[indy] == '^': #hit a splitter, add the split, add new locations of split beams
            hit_splitters +=1
            new_locs = new_locs | {indy-1, indy+1,}
        else: #beam not hit a splitter, pass its index to the next line
            new_locs.add(indy)
    return hit_splitters, new_locs
    
def count_all_splits(lines:list[str])-> int:
    '''Solves part1 by looping through text input.
    lines: list of str text input
    Returns: all_splits: integer, number of times beams hit splitters'''
    all_splits = 0
    beam_locs= {lines[0].index('S'),}
    for i in range(2, len(lines),2): #splitters appear every 2 lines
        new_splits, beam_locs = new_beams_per_line(lines[i], beam_locs)
        all_splits += new_splits
        # print(i, beam_locs, all_splits)
    return all_splits
```

```python
#Part 2 Functions

def new_particle_routes_per_line (newline:str, locs_dict:dict[int:int]) -> (int, set[int]):
    '''Helper function for Part 2 task.
    newline : str of new line with splittes ^ and spaces .
    particle_locs: dictonary of incoming beam location (str index) as key and the number of incoming routes
        through that index 
    Returns: 
        new_locs: dictonary of outgoing beam location (str index) as key and the number of outgoing routes
        through that index 
        ''' 
    
    new_locs = {}
    for indy in locs_dict: #indices of incoming beams
        if newline[indy] == '^': #hit a splitter, add the route through new beam locations
            new_locs[indy+1] = new_locs.get(indy+1,0)+locs_dict[indy]
            new_locs[indy-1] = new_locs.get(indy-1,0)+locs_dict[indy]
        else: #beam not hit a splitter, pass its index to the next line
            new_locs[indy] = new_locs.get(indy,0) + locs_dict[indy]
    return new_locs

def count_all_routes(lines:list[str]) -> int:
    '''Part 2 counts all the possible routes (they say timelines, but they are really routes)
    of the particle through the tachyon manifold.
    This function loops through the input text (the manifold), running the function
    new_particle_routes_per_line to keep track of all new particle locations and the number of
    routes that traverse each point.

    At the end, the routes through all ending particle locations are added up and returned.
    lines: list of str text input
    Returns: total_routes: integer, all possible routes of a single particle through the
        tachyon manifold'''

    particle_locs= {lines[0].index('S'):1} #first route
    for i in range(2, len(lines),2): #splitters appear every 2 lines
        particle_locs = new_particle_routes_per_line(lines[i], particle_locs)
    total_routes = sum(particle_locs.values())
    return total_routes


```
