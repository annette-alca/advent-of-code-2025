Solution to [Day 7 Laboratories. Tachyon Manifold Beams](https://adventofcode.com/2025/day/7).

Part 1. Count the number of times the beam is split down the path. Part 1 uses *new_beams_per_line* to keep track of beam locations and the number of times a beam is split for each new line (or row) of data. The function *count_all_splits* loops through the data to count the splits.

-AA



```python
#Part 1 Functions

from typing import Any

def new_beams_per_line(newline:str, beam_locs:set[int]) -> (int, set[int]):
    '''Helper function for the task.
    newline : str of new line with splittes ^ and spaces .
    beam_locs: set of integers, using sets to give only unique indices
    Returns: hit_splitters : int, number of splitters the incoming beams hit
        new_locs: set of integers, unique indices of outgoing beams
        ''' 
    hit_splitters = 0
    new_locs = set[int]()
    for indy in beam_locs: #indices of incoming beams
        if newline[indy] == '^': #hit a splitter, add the split, add new locations of split beams
            hit_splitters +=1
            new_locs = new_locs| {indy-1, indy+1,}
        else: #beam not hit a splitter, pass its index to the next line
            new_locs.add(indy)
    return hit_splitters, new_locs
    
def count_all_splits(lines:list[str])-> int:
    '''Solves part1 by looping through text input.
    lines: list of str text input
    Returns: all_splits: integer, number of times beams hit splitters'''
    all_splits = 0
    beam_locs= {lines[0].index('S'),}
    for i in range(2, len(lines),2): #splitters appear every 2 lines
        new_splits, beam_locs = new_beams_per_line(lines[i], beam_locs)
        all_splits += new_splits
        # print(i, beam_locs, all_splits)
    return all_splits
```

```python
#Part 1 Sample

sampletext = '''.......S.......
...............
.......^.......
...............
......^.^......
...............
.....^.^.^.....
...............
....^.^...^....
...............
...^.^...^.^...
...............
..^...^.....^..
...............
.^.^.^.^.^...^.
...............'''
s_lines = sampletext.split('\n')
count_all_splits(s_lines)
```
