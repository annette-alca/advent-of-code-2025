Solution to [Day 9 Largest Rectangle](https://adventofcode.com/2025/day/9).

Similar to Day 8, Day 9 gives us a list of coordinates (but this time to a 2-dimensional matrix). Part 1 asks for the area of the largest rectangle, solved here using *get_max_area*. The function *helper_convert_lines_to_list_coords* is reused from Day 8 (that I have not pushed yet).

-Annette

```python

def helper_convert_lines_to_list_coords(lines:list[str]) -> list[list]:
    str_list = [l.split(',') for l in lines]
    coords_list = [[int(e) for e in l] for l in str_list]
    return coords_list

def get_max_area(lines: list[str]) -> int:
    '''
    This function loops through all the coordinates to calculate areas for all pairs.
    Areas are solved length * width, maximum area is returned
    '''
    input_list = helper_convert_lines_to_list_coords(lines)
    all_areas = [] 

    for i in range(len(input_list)-1):
        x1,y1 = input_list[i]
        for j in range(i+1, len(input_list)):
            x2,y2 = input_list[j]
            area = abs((x1-x2+1)*(y1-y2+1)) #1 is added to the length and widths as they are inclusive of both points
            all_areas.append(area)
    return max(all_areas)
```

```python
#Part 1 Sample. Expected Answer is 50.

sampletext = """7,1
11,1
11,7
9,7
9,5
2,5
2,3
7,3"""
slines = sampletext.split('\n')
get_max_area(slines)
```

```python
#Part 1 Solution

with open('inputs/day9_input.txt', 'r') as f:
    lines = f.read().splitlines()

print(f"Part 1 Answer, area of largest table: {get_max_area(lines)}")
```
