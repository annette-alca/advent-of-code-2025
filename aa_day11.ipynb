Solution to [Day 11. Reactor. Counting Paths](https://adventofcode.com/2025/day/11).

Part 1. Counting Possible Paths from 'you' to 'out'. This reminds me of Day 7, the Tachyon Manifold, except for the fact that the input lines are not in order. I solved this in a similar way, using intermediate steps (and how many paths go through those steps) until the end is reached. The steps are done through function *helper_viable_paths* used in a loop with function *paths_from_you*. I worked backwards from all instances of 'out' until I ended up with 'you'.

-Annette

```python

def helper_convert_lines_to_dict(lines:list[str]) -> dict:
    '''convert input file to dictionaries'''
    str_list = [line.split(':') for line in lines]
    # print(str_list)
    path_d = {l[0]:l[1].strip().split(' ') for l in str_list}
    return path_d

def helper_viable_paths(path_d:dict, temp_at: dict = {}, path_count:int=0)-> (list, int):
    '''does one step back in the path
    path_d: dictionary of strings:list, from the input file
    temp_at: dictionary of strings:int, each string is a step in the path and the int is how many routes go through that step
    path_count: paths that already ended at you
    Returns: new_at, strings:int next group of strings working backwards and how many routes go through each one'''
    if not temp_at:
        new_keys = [k for k in path_d.keys() if 'out' in path_d[k]]
        return {k:1 for k in new_keys}, 0
    else:
        new_at = {}
        try:
            path_count += temp_at.pop('you') #removes steps at you, and takes number of paths at that point
        except KeyError:
            # print(path_count)
            pass
        for t,val in temp_at.items():
            new_keys = [k for k in path_d.keys() if t in path_d[k]]
            # print('>',t, new_keys)
            for k in set(new_keys):
                new_at[k] = new_at.get(k,0) + new_keys.count(k)*val
    return new_at, path_count        

def paths_from_you(lines: list[str])-> int:
    '''Loops through helper_viable_paths to solve Part 1
    Loops stop when there are no more step backwards available.
    lines: text input 
    Returns: int, number of paths from 'out' backwards to you.'''
    path_d = helper_convert_lines_to_dict(lines)
    temp_at = None
    path_count = 0
    while temp_at != {}: #out of steps
        temp_at, path_count = helper_viable_paths(path_d, temp_at, path_count)
        # print(temp_at, path_count)
    return path_count
    
```

```python
#Part 1 Sample
sampletext = """aaa: you hhh
you: bbb ccc
bbb: ddd eee
ccc: ddd eee fff
ddd: ggg
eee: out
fff: out
ggg: out
hhh: ccc fff iii
iii: out"""
slines = sampletext.split('\n')
print (f"Part 1 sample answer = {paths_from_you(slines)} paths.")
```
