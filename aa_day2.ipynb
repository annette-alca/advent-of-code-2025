Part 1. Find pattern of twice repeated sequence of digits in a number. In previous Advent of Codes, Part 1 can be done by brute force, which is what I'm doing here without knowing what Part 2 will ask. The approach checks every number in the range to see if it is a valid id and sums all valid ids.

```python
#Part 1 Functions

def valid_id(num:int):
    '''Determine if a number is a valid id following Part 1 definition.
    This requires the number to have an event number of digits.
    Returns: number if it is a valid id, 0 if not'''
    lenstr = len(str(num))
    if lenstr%2 == 1:
        return 0
    else:
        half = lenstr/2
        if num//10**half == num%(10**half):
            return num
        else:
            return 0

def sum_valid_ids(numstr: str) -> int:
    '''Converts string into a range and sums all valid ids''''
    start, end = numstr.split('-')
    sum_valids = 0
    for num in range(int(start), int(end)+1):
        sum_valids += valid_id(num)
    return sum_valids


```

```python
# Part 1 Sample
sample = '11-22,95-115,998-1012,1188511880-1188511890,222220-222224,1698522-1698528,446443-446449,38593856-38593862,565653-565659,824824821-824824827,2121212118-2121212124'.split(',')
sum([sum_valid_ids(s) for s in sample])
```

```python
# Part 1 Answer

with open('inputs/day2_input.txt','r') as f:
    inputline = f.readline()
    # print(len(inputline))
inputlist = inputline.split(',')
print(f'Part 1 Answer, sum of all valid ids = {sum([sum_valid_ids(i) for i in inputlist])}')
```
