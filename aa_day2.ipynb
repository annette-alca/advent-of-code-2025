Solution to [Day 2 Gift Shop](https://adventofcode.com/2025/day/2).

Part 1. Find pattern of twice repeated sequence of digits in a list of numbers. In previous Advent of Codes, Part 1 can be done by brute force, which is what I'm doing here without knowing what Part 2 will ask. The approach checks every number in the range to see if it is an invalid id (fits the pattern) and sums all invalid ids. Functions used are called *invalid_id* and *sum_invalid_ids*.

Part 2. It turns out this can be also be solved by brute force. There are additional patterns allowed for valid ids, defined in the *invalid_id_pt2* function. There is also the *sum_invalid_ids_pt2* function and two helper functions to assess numbers for their invalidity.

```python
#Part 1 Functions

def invalid_id(num:int):
    '''Determine if a number is an invalid id following Part 1 definition.
    This requires the number to have an event number of digits.
    Returns: number if it is an invalid id, 0 if not'''
    lenstr = len(str(num))
    if lenstr%2 == 1:
        return 0
    else:
        half = lenstr/2
        if num//10**half == num%(10**half):
            return num
        else:
            return 0

def sum_invalid_ids(numstr: str) -> int:
    '''Converts string into a range and sums all valid ids'''
    start, end = numstr.split('-')
    sum_valids = 0
    for num in range(int(start), int(end)+1):
        sum_valids += invalid_id(num)
    return sum_valids


```

```python
# Part 1 Sample
sample = '11-22,95-115,998-1012,1188511880-1188511890,222220-222224,1698522-1698528,446443-446449,38593856-38593862,565653-565659,824824821-824824827,2121212118-2121212124'.split(',')
sum([sum_invalid_ids(s) for s in sample])
```

```python
# Part 2 Functions
import math

def helper_get_divisors(bignum:int) -> set[int]:
    '''Provide all divisors of a big number.
    bignumber: decimal places of a number
    Returns:set of all possible lengths of repeats'''
    divisors = []
    for i in range(2, math.floor(bignum**0.5) + 1):
        if bignum%i==0:
            divisors.append(i)
            divisors.append(bignum//i)
            # print(divisors)
    return (set(divisors))

def helper_all_repeats(numstr:str, len_all:int, len_repeat:int)->bool:
    '''Tests whether a number has perfect repeats of a given unit length of repeat.
    Returns True or False
    e.g. helper_all_requests('123123',3) -> True
    but helper_all_requests('123123',2) -> False'''
    num_repeats = len_all//len_repeat
    firstrep = numstr[:len_repeat]
    for i in range(1, num_repeats): 
        if numstr[i*len_repeat:(i+1)*len_repeat]!=firstrep:
            return False
    return True
    

def invalid_id_pt2(num: int) -> int:
    '''Determine if a number is an invalid id. All perfect repeats allowed.
    This is simpler to solve as a string than as integer.
    Returns: number if it is an invalid id, 0 if not'''
    numstr = str(num)
    uniques = set(numstr)
    lu = len(uniques)
    la = len(numstr) # short for length of all
    if la==1:
        return 0
    elif lu== 1: #single repeating number
        return num
    elif lu == la: #no repeats
        return 0
    else:
        len_repeat_candidates = sorted(helper_get_divisors(la), reverse=True)  #decreasing repeat size
        # print(len_repeat_candidates)
        for lr in len_repeat_candidates:
            if helper_all_repeats(numstr, la,lr):
                return num
        return 0

def sum_invalid_ids_pt2(numstr: str) -> int:
    '''Converts string into a range and sums all valid ids'''
    start, end = numstr.split('-')
    sum_valids = 0
    for num in range(int(start), int(end)+1):
        sum_valids += invalid_id_pt2(num)
    return sum_valids
```
