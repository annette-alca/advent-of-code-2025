Part 1. Safe combination

```python
def split_rotation(move:str) -> int:
    dir = move[0]
    num = int(move[1:])%100
    if dir == 'L':
        return -num
    elif dir == 'R':
        return num
        print(num)
    else:
        raise ValueError(f"Invalid direction: {dir}")

def safe_combination(sequence:list[str]) -> int:
    """
    Add all rotations of a sequence of strings
    """
    spot = (50 + sum([split_rotation(s) for s in sequence])) % 100
    start = 50
    password = 0
    for s in sequence:
        start += split_rotation(s)
        start = start%100
        print (start)
        if start == 0:
            password += 1
    return password
    # if spot>+0:
    #     return spot
    # else:
    #     return 100 - spot
```

```python
sample = '''L68
L30
R48
L5
R60
L55
L1
L99
R14
L82'''.split('\n')
safe_combination(sequence=sample)
```
