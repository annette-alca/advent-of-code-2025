Solution to [Day 1](https://adventofcode.com/2025/day/1)

Part 1. Lock with combination dial going from 0-99. Answer how many times the series of moves stops at zero.

Part 2. Same lock. In addition to Part 1 answer, include how many times the series of moves pass zero. 

-Annette

```python
# Part 1 Functions
def split_move(move:str) -> int:
    """split move into number of clicks and direction, give move as a positive or negative number"""
    dir = move[0]
    num = int(move[1:])%100 #because lock only has 0-99
    if dir == 'L':
        return -num
    elif dir == 'R':
        return num
    else:
        raise ValueError(f"Invalid direction: {dir}")

def lock_at_zero(sequence:list[str]) -> int:
    """
    Do each move/rotation and count number of times dial stops at zero
    """
    spot = 50
    password = 0
    for s in sequence:
        spot += split_move(s)
        spot = spot%100 #because lock only has 0-99
        if spot == 0:
            password += 1
    return password

```

```python
# Part 1 Sample Test
sample = '''L68
L30
R48
L5
R60
L55
L1
L99
R14
L82'''.split('\n')
lock_at_zero(sequence=sample)
```

```python
# Part 1 Answer
with open('inputs/day1_input.txt', 'r') as file:
    sequence = file.read().splitlines()

print(f"Password is {lock_at_zero(sequence)}")
```

```python
# Part 2 Functions
def split_move_pt2(move:str) -> (int, int):
    """split move into number of clicks and direction (L or R) and give a positive or negative number;
    also give number of full revolutions i.e. if move>100"""
    dir = move[0]
    raw_num = int(move[1:])
    num = raw_num%100 #because lock only has 0-99
    revolutions = raw_num//100
    if dir == 'L':
        return -num, revolutions
    elif dir == 'R':
        return num, revolutions
    else:
        raise ValueError(f"Invalid direction: {dir}")

def lock_pass_or_stop_at_zero(sequence:list[str]) -> int:
    """
    Do each move/rotation and count number of times dial stops at zero or passes zero
    """
    spot = 50
    password = 0
    at_zero = False

    for s in sequence:
        move, revolutions = split_move_pt2(s)
        spot += move
        password += revolutions
        #dial passes zero
        if spot>100:
            password +=1
        if spot<0 and not at_zero: #turning left from >0
            password +=1

        #get actual number 0-99
        spot = spot%100
        
        #dial stops at zero
        if spot == 0:
            password += 1
        at_zero = spot == 0
    return password
```

```python
#Part 2 Sample Test
print(lock_pass_or_stop_at_zero(sample))
```

```python
#Part 2 Answer
print(f"Password is {lock_pass_or_stop_at_zero(sequence)}")
```
